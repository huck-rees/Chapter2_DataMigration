```python
import os
import pandas as pd

# ==============================================================================
# USER CONFIGURATION
# ==============================================================================

CONFIG = {
    "master_csv": r"E:\Dissertation\Data\GreenbergZhaoRees_river_datasheet.csv",
    "output_csv": r"E:\Dissertation\Data\compiled_GACP_classifications.csv",
}

# ==============================================================================
# COMPILE
# ==============================================================================

river_data = pd.read_csv(CONFIG["master_csv"])
records = []

for _, row in river_data.iterrows():
    river_name    = row["river_name"]
    working_dir   = row["working_directory"]

    csv_path = os.path.join(
        working_dir, "Luo_etal_2026", "Extracted_Rivers",
        river_name, f"{river_name}_GACP_classifications.csv"
    )

    if not os.path.exists(csv_path):
        print(f"Missing: {river_name}")
        continue

    df = pd.read_csv(csv_path)
    records.append(df[["river_name", "ds_order", "GACP_class_int", "GACP_class"]])

compiled = pd.concat(records, ignore_index=True)
compiled.to_csv(CONFIG["output_csv"], index=False)

print(f"\nCompiled {len(compiled)} reaches across {compiled['river_name'].nunique()} rivers.")
print(f"Saved to: {CONFIG['output_csv']}")
print(f"\nClass distribution:")
print(compiled["GACP_class"].value_counts().to_string())
compiled.head(10)
```
