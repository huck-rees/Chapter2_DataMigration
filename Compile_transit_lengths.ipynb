```python
import os
import pandas as pd

# ==============================================================================
# USER CONFIGURATION
# ==============================================================================

CONFIG = {
    "master_csv": r"E:\Dissertation\Data\GreenbergZhaoRees_river_datasheet.csv",
    "output_csv": r"E:\Dissertation\Data\RiverMapping\Mobility\compiled_transit_lengths.csv",
}

# ==============================================================================
# COMPILE
# ==============================================================================

river_data = pd.read_csv(CONFIG["master_csv"])
records = []

for _, row in river_data.iterrows():
    river_name  = row["river_name"]
    working_dir = row["working_directory"]

    csv_path = os.path.join(
        working_dir, "RiverMapping", "Mobility",
        river_name, f"{river_name}_transit_lengths.csv"
    )

    if not os.path.exists(csv_path):
        print(f"Missing: {river_name}")
        continue

    df = pd.read_csv(csv_path)
    df["river_name"] = river_name
    records.append(df[["river_name", "ds_order", "real_reach_length_m", "x_tran_m", "n_stor"]])

compiled = pd.concat(records, ignore_index=True)
compiled.to_csv(CONFIG["output_csv"], index=False)

print(f"Compiled {len(compiled)} reaches across {compiled['river_name'].nunique()} rivers.")
print(f"Columns: {compiled.columns.tolist()}")
print(f"Saved to: {CONFIG['output_csv']}")
compiled.head(10)
```
